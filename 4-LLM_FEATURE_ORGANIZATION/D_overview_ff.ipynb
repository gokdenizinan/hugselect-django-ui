```python
# 1️⃣ Count rows where Base_author and Base_model are both "yes"
yes_both = df_feature[(df_feature['Base Author'] == 'yes') | (df_feature['Base Model'] == 'yes')].shape[0]

# 2️⃣ Out of the ones that AREN'T both "yes"
not_yes_both = df_feature[~((df_feature['Base Author'] == 'yes') | (df_feature['Base Model'] == 'yes'))]
not_yes_both_count = not_yes_both.shape[0]

# 3️⃣ Of those "not yes" rows, count how many have Technical == "yes"
tech_yes = not_yes_both[not_yes_both['Technical'] == 'yes'].shape[0]

# 4️⃣ Out of the ones that DON'T have Technical == "yes",
# count how many have Characteristic == "yes" and Functional == "yes"
not_tech_yes = not_yes_both[not_yes_both['Technical'] != 'yes']
char_func_yes = not_tech_yes[
    (not_tech_yes['Characteristic'] == 'yes') |
    (not_tech_yes['Functionality'] == 'yes')
].shape[0]

# 5️⃣ Print everything neatly
print(f"Rows where Base_author and Base_model are 'yes': {yes_both}")
print(f"Rows that are NOT both 'yes': {not_yes_both_count}")
print(f"Within those, rows where Technical is 'yes': {tech_yes}")
print(f"Of the remaining, rows where Characteristic and Functional are 'yes': {char_func_yes}")
```
